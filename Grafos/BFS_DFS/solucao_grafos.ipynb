# Solução – Grafos, BFS e DFS

**Disciplina:** Estrutura de Dados II – Ciência da Computação

## 1. Importações e Modelagem do Grafo

```python
from collections import deque
from typing import Dict, List, Optional

# Grafo não dirigido com 9 vértices (lista de adjacência)
GRAFO: Dict[str, List[str]] = {
    "Brasília": ["Goiânia", "Belo Horizonte", "Cuiabá"],
    "Goiânia": ["Brasília", "São Paulo", "Cuiabá", "Campo Grande"],
    "Belo Horizonte": ["Brasília", "Rio de Janeiro", "São Paulo"],
    "São Paulo": ["Goiânia", "Belo Horizonte", "Curitiba", "Rio de Janeiro"],
    "Cuiabá": ["Brasília", "Goiânia", "Campo Grande", "Curitiba"],
    "Rio de Janeiro": ["Belo Horizonte", "São Paulo", "Curitiba"],
    "Curitiba": ["São Paulo", "Cuiabá", "Rio de Janeiro", "Florianópolis"],
    "Campo Grande": ["Goiânia", "Cuiabá"],
    "Florianópolis": ["Curitiba"],
}

print("Grafo carregado com", len(GRAFO), "vértices.")
print("\nLista de Adjacência:")
for vertice in sorted(GRAFO.keys()):
    print(f"  {vertice}: {GRAFO[vertice]}")
```

## 2. Implementação do BFS (Busca em Largura)

```python
def bfs(rede: Dict[str, List[str]], origem: str, destino: str) -> Optional[List[str]]:
    """Retorna caminho com menor número de arestas (grafo não ponderado)."""
    if origem not in rede or destino not in rede:
        return None

    fila = deque([(origem, [origem])])
    visitados = {origem}

    while fila:
        atual, caminho = fila.popleft()

        if atual == destino:
            return caminho

        for vizinho in rede.get(atual, []):
            if vizinho not in visitados:
                visitados.add(vizinho)
                fila.append((vizinho, caminho + [vizinho]))

    return None

print("Função BFS definida.")
```

## 3. Implementação do DFS (Busca em Profundidade)

```python
def dfs(rede: Dict[str, List[str]], origem: str, destino: str) -> Optional[List[str]]:
    """Retorna um caminho qualquer (não necessariamente o menor)."""
    if origem not in rede or destino not in rede:
        return None

    pilha = [(origem, [origem])]
    visitados = set()

    while pilha:
        atual, caminho = pilha.pop()

        if atual in visitados:
            continue
        visitados.add(atual)

        if atual == destino:
            return caminho

        for vizinho in reversed(rede.get(atual, [])):
            if vizinho not in visitados:
                pilha.append((vizinho, caminho + [vizinho]))

    return None

print("Função DFS definida.")
```

## 4. Função de Bloqueio de Conexão

```python
def bloquear_conexao(rede: Dict[str, List[str]], a: str, b: str) -> Dict[str, List[str]]:
    """Cria cópia do grafo e remove a aresta a–b nos dois sentidos."""
    copia = {no: vizinhos.copy() for no, vizinhos in rede.items()}

    if b in copia.get(a, []):
        copia[a].remove(b)
    if a in copia.get(b, []):
        copia[b].remove(a)

    return copia

print("Função bloquear_conexao definida.")
```

## 5. Funções auxiliares

```python
def formatar_caminho(caminho: Optional[List[str]]) -> str:
    if caminho is None:
        return "Nenhum caminho encontrado"
    arestas = len(caminho) - 1
    return f"{' → '.join(caminho)}  ({arestas} aresta{'s' if arestas != 1 else ''})"

def executar_busca(rede: Dict[str, List[str]], origem: str, destino: str, nome_cenario: str) -> None:
    print(f"\n--- Cenário: {nome_cenario} ---")
    print(f"Origem: {origem}  |  Destino: {destino}")

    caminho_bfs = bfs(rede, origem, destino)
    caminho_dfs = dfs(rede, origem, destino)

    print(f"BFS: {formatar_caminho(caminho_bfs)}")
    print(f"DFS: {formatar_caminho(caminho_dfs)}")

print("Funções auxiliares definidas.")
```

## 6. Testes Automáticos

### Experimento A – Rede completa

```python
executar_busca(GRAFO, "Brasília", "Curitiba", "Rede completa (sem bloqueios)")
```

### Experimento B – Bloqueio Goiânia ↔ São Paulo

```python
rede_bloqueada1 = bloquear_conexao(GRAFO, "Goiânia", "São Paulo")
executar_busca(rede_bloqueada1, "Brasília", "Curitiba", "Bloqueio: Goiânia ↔ São Paulo")
```

### Experimento C – Bloqueio adicional Cuiabá ↔ Curitiba

```python
rede_bloqueada2 = bloquear_conexao(rede_bloqueada1, "Cuiabá", "Curitiba")
executar_busca(rede_bloqueada2, "Brasília", "Curitiba", "Bloqueio adicional: Cuiabá ↔ Curitiba")
```

### Caminho até Florianópolis (folha)

```python
executar_busca(GRAFO, "Brasília", "Florianópolis", "Caminho até Florianópolis")
```

### Isolamento de Florianópolis

```python
rede_isolada = bloquear_conexao(GRAFO, "Curitiba", "Florianópolis")
executar_busca(rede_isolada, "Brasília", "Florianópolis", "Bloqueio Curitiba ↔ Florianópolis (isola a folha)")
```

Altere as variáveis `origem` e `destino` abaixo e execute a célula.

## 8. Simular um bloqueio personalizado

Altere as cidades `cidade_a` e `cidade_b` para simular o bloqueio de outra aresta.

```python
cidade_a = "Goiânia"
cidade_b = "São Paulo"
origem = "Brasília"
destino = "Curitiba"

rede_temp = bloquear_conexao(GRAFO, cidade_a, cidade_b)
executar_busca(rede_temp, origem, destino, f"Bloqueio temporário: {cidade_a} ↔ {cidade_b}")
```

## 9. Observações finais

- **BFS** encontra o caminho com o **menor número de arestas** (grafo não ponderado).
- **DFS** encontra **um** caminho válido, mas a ordem dos vizinhos influencia o resultado.
- A função `bloquear_conexao` cria uma **cópia**, preservando o grafo original.
- Esta é uma simulação **didática** — não deve ser usada para orientar voos reais.
